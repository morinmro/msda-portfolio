### C1. Plan to Assess Quality of Data

My plan for assessing data quality in the Churn dataset involves a multi-step approach that will comprehensively identify issues within the data. This will involve identifying duplicates, missing or null data, and outliers. I will also evaluate whether re-expression of categorical variables is required to standardize column formats for boolean values.

Duplicate rows will be identified based on two factors: complete duplicate rows, which contain matching data across all columns; and unique-value duplicates, which contain duplicate values on columns that should be unique to reach record. The columns used for the latter will be those which should be unique as defined in the Churn data dictionary; these are: Case Order, Customer ID, and Interaction. In both cases, the duplicated() method will be used to find the duplicates. By default, duplicated() will return complete duplicates. Using the subset argument for duplicated(), the duplicate check can be narrowed to specific columns, such as the ones listed above. (Pandas Documentation, 2024)

The next step will involve identifying missing data – that is, records that have blank or null values for any column. Total null values can be found using a combination of the Python methods isnull() and sum(), which will return a list of all columns along with the total null values for each.

Outliers in quantitative data will be identified by generating boxplots, which can be visually inspected in order to determine if any values fall outside of the column’s expected range. This will be done using the Seaborn library. Boxplots will be created for the following quantitiative fields: Age, Tenure, Bandwidth_GB_Year, Income, Children, Email, Contacts, and Yearly_equip_failure. Once the boxplots have been generated to visually show the data, the query() Python method will be used to get counts of data points outside of the range indicated by the boxplot whiskers.

For categorical data, out-of-range values will be identified by printing lists of unique values for each column and inspecting these lists to ensure that no unexpected values are present. This applies to the following fields: Education, Employment, Marital, Gender, Churn, Techie, Contract, Port_modem, Tablet, InternetService, Phone, Multiple, OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport, StreamingTV, StreamingMovies, PaperlessBilling, PaymentMethod, item1, item2, item3, item4, item5, item6, item7, and item8.

Additionally, I will evaluate all columns with expected boolean values to ensure that they use a common format (e.g, that they are all either Yes/No, or 0/1). This applies to the following columns: Churn, Techie, Port_modem, Tablet, Phone, Multiple, OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport, StreamingTV, StreamingMovies, and PaperlessBilling.

### C2. Justification of Approach

Identifying complete duplicates, as well as duplicate values across unique columns, will help us ensure quality of the data being analyzed, and prevent our resulting analysis from being incorrectly skewed by repetitions within the data. The columns that we are inspecting for duplicates are those that are explicitly defined as having unique values in each observation, based on the Churn data dictionary that accompanied the dataset. In both cases, the duplicated() method can accurately identify these duplicates. (Pandas Documentation, 2024)

Missing data is easy to spot through a variety of methods, but using isnull() and sum() together will return the total number of missing values in each column.

Using boxplots to inspect quantitative data is an effective method of quickly identifying outliers. Outliers are shown as points outside of the whiskers of the boxplot, making them easy to spot at a glance. The query() method, in combination with info(), will allow for easily counting the number of outliers for each column; query() can be used to select only the outliers and place them into new dataframes, and info() will print summaries of these new dataframes, including counts.

Values in categorical columns must fall within expected ranges (e.g., values in item1 must fall along an eight-point scale from 1 to 8 - a value of 9 would be considered out of range). Printing lists of unique values contained within the dataset for each of these columns will allow for the identification of any values that fall outside of the expected ranges.

Ordinal data with boolean values must have their formats standardized in order for the data to be considered clean. The preference is to have all such fields reflect either Yes/No or 0/1. The unique() method in Python will return lists of unique values for each of these fields, allowing us to verify that all fields are formatted in a standardized way. (Pandas Documentation, 2024)

Taken together, all of these steps will comprehensively identify data issues in the Churn dataset; the findings resulting from this data quality assessment can be used to formulate mitigation methods for each issue that is identified.

### D2. Justification of Mitigation Methods

The unnamed column appearing at the beginning of the dataset is not mentioned in the data dictionary for this file, and it contains no data that is useful or relevant to our usecase. As such, this column will be dropped.

The missing values in each field will be addressed on a case-by-case basis, with an appropriate cleaning method applied based on the distribution of the data. Missing values in the Children field will be replaced with zeroes; it is reasonable to assume that a missing or null value in this field reflects an absence of children in the household. Missing values in the Age and Income fields will be replaced with the median values for each; this prevents introduction of skewness due to high outlier values. Tenure and Bandwidth_GB_Year will have their missing values replaced with the mean, as these are already normal distributions with no outliers present. Techie, Phone, and TechSupport will have all missing values defaulted to 'No', as it is reasonable to assume that a respondant who did not answer this question with an explicit affirmative does not fall into these categoriges (that is, either does not subscribe to the relevant service, or does not self-identify as technically adept).

No explicit action will be taken on the outliers detected in the Income, Children, Contacts, Email, and Yearly_equip_failure fields. The outliers that are present, while significantly higher than the mean or median values for these fields, are not outside of the realm of reason. For example, it is entirely possible that family has 10 children in the household, or that a household's yearly income exceeding $250,000. Put another way, while these values are indeed high relative to the entire population, they are not impossible. Removing or replacing these values or records could have an adverse affect on the dataset as a whole and could deprive the analysis of insights relating to high-income families, or large families. The same goes for Contacts and Yearly_equip_failure; removing these outliers would actually be counter-productive as it would hamper any attempts at analyzing the data in regards to the most trouble-prone customers (e.g., those who have reason for repeated contacts with customer support, or those who have had a higher-than-average rate of equipment trouble).

Finally, the Education column will be re-expressed into numeric values; these values will be stored in a new column, Education_Num. The Education column is ordinal, meaning that each individual value within the column can be said to be positioned either "higher" or "lower" than any other value - there is an implicit heirarchy to the values stored in this column, and therefore these values can be re-expressed on a numerical scale. This is not, strictly speaking, a data cleaning step - rather, it falls under the category of data wrangling; however, performing this re-expression will structure the data in preparation for for future modeling.

### D4. Mitigation Code

```python
# Drop Unnamed column

churnfile.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# Replace missing values

# Children: replace NA with 0
# It is reasonable to assume that a missing value for Children
# reflects the absence of children in the household
churnfile['Children'] = churnfile['Children'].fillna(0)

# Age, Income: replace NA with median
# Using median for income prevents introduction of skewness due to higher values
for col in ('Age','Income'):
    churnfile[col] = churnfile[col].fillna(churnfile[col].median())

# Tenure, Bandwidth_GB_Year: replace with mean
for col in ('Tenure','Bandwidth_GB_Year'):
    churnfile[col] = churnfile[col].fillna(churnfile[col].mean())
    
# Techie, Phone, TechSupport: default to No
# It is reasonable to assume that a missing value for any of these fields
# reflects a negative answer
for col in ('Techie','Phone','TechSupport'):
    churnfile[col] = churnfile[col].fillna('No')
    
```

```python
# Re-expression of Education column
# This field is ordinal; that is, all values in Education can be said to be either
# "Above" or "Below" all other values within this column.
# Therefore, we can assign a numeric order to these values, with 0 reflecting no education,
# and 11 reflecting a doctorate.


# First, begin by reviewing the unique values for this field
print('Unique values in Education: ')
print(churnfile['Education'].unique())

# Create a new variable to store the numeric values
churnfile['Education_Num'] = churnfile['Education']

# Then, establish a dictionary to map values to numbers
dict_education = {"Education_Num":{"No Schooling Completed":0,"Nursery School to 8th Grade":1,"9th Grade to 12th Grade, No Diploma":2,
           "GED or Alternative Credential":3,"Regular High School Diploma":4,"Some College, Less than 1 Year":5,
           "Some College, 1 or More Years, No Degree":6,"Associate's Degree":7,"Bachelor's Degree":8,
            "Master's Degree":9,"Professional School Degree":10,"Doctorate Degree":11}}

# Use the dictionary to populate values in Education_Num
churnfile.replace(dict_education, inplace=True)

# Print unique vaues in Education_Num to verify field populated correctly
print('Unique values in Education_Num: ')
print(churnfile['Education_Num'].unique())
```

### D6. Limitations

Although the cleaning actions performed on this dataset have thoroughly addressed the issues present in the data, these actions cannot be considered as a perfect solution to these issues; indeed, without revised and accurate data directly from the data source, all steps taken to correct and clean the data must be considered as best-effort measures that come with their own caveats and pitfalls. No data cleaning steps will ever be as accurate as receiving complete and correct data in the first place, as many assumptions must be made in data cleaning that could be incorrect.

Any analysis performed on the newly-cleaned Churn dataset must be done with an understanding as to the limitations introduced by the cleaning process. Missing values were populated based on best guesses, taking into consideration the nature of each column and their respective real-world contexts. It is entirely possible, for example, that missing values for the Children column actually reflect a real value other than zero that could not be communicated correctly in the file we received for analysis; perhaps the initial survey that collected this data had some records that recorded this field as text rather than as numeric whole numbers (e.g., "two" rather than "2"), and these values were erroneously discarded during the data acquisition phase as the data analyst importing the original survey data was expecting only numbers for this field. This would mean that our approach to cleaning the Children column unintentionally introduced bad data into the dataset, which would make further analysis inaccurate. This is true for all of the columns that had their null and missing values populated: Children, Age, Income, Techie, Phone, TechSupport, Tenure, and Bandwith_GB_Year. While the specific values we introduced in each column to replace the nulls were different, the overall concern is the same: that we may have inadvertently introduced inaccurate data into the dataset, thereby skewing the data and decreasing the accuracy of future analyses.

The same is true in the case of outliers, which in this case were kept in the dataset as the assumption was made that these represent accurate data. It is possible that this assumption was incorrect, and the distribution for some columns is therefore incorrectly skewed; it could be that a value of 10 in the Children column was actually meant to be 1, and the Children column is erroneously being skewed to the right because of these outlying values.

The impacts described above apply to all columns affected by the mitigation code provided previously; any column that had null values populated with derived values, or outlier values that were not discarded, could potentially skew the data and affect the analytic outcomes. Analysts working with this dataset, as well as business stakeholders, must be made aware of these limitations and how they could affect an analysis.


### E1. Principal Components

```python
# Import necessary packages for PCA
# We already have pandas and seaborn as they were imported earlier

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Define variables we'll us for PCA
# These variables are all continuous numerical variables, as discrete or categorical variables
# should not be used for PCA.
churn_pca = churnfile[['Income','Tenure','MonthlyCharge','Bandwidth_GB_Year','Outage_sec_perweek']]

# Normalize data; standardization/mean normalization
# Data minus mean, divided by standard deviation
# Prevents one variable from being overly influential
churn_pca_norm = ((churn_pca-churn_pca.mean())/churn_pca.std())

# Shape data on number of principle components created
# First, grab number of PCs that will be created
pca = PCA(n_components=churn_pca.shape[1])
print(pca)
```

```python
#Then, fit the data
pca.fit(churn_pca_norm)

churn_pca_final = pd.DataFrame(pca.transform(churn_pca_norm), columns=['PC1','PC2','PC3','PC4','PC5'])
```

```python
# Get loadings
loadings = pd.DataFrame(pca.components_.T,
                             columns=['PC1','PC2','PC3','PC4','PC5'], index=churn_pca.columns)

loadings
```

### E2. Criteria Used

Five PCs were generated as part of the above code. In selecting which values to keep, I'll reference the Kaiser Rule, which states that PCs with eigenvalues greater than 1 should be kept while others should be discarded. First I'll calculate the eigenvalues for each PC, then use matplotlib to visualize these on a scree plot.

```python
# Calculate covariance and vectors, then define eigenvalues
covar_mtx = np.dot(churn_pca_norm.T, churn_pca_norm) / churn_pca.shape[0]
eigenvals = [np.dot(eigenvector.T, np.dot(covar_mtx, eigenvector)) for eigenvector in pca.components_]

# Create scree plot
plt.plot(np.arange(1,len(eigenvals)+1),eigenvals,marker='o')
plt.xlabel('Numer of Components')
plt.ylabel('Eigenvalue')
plt.axhline(y=1, color="red")
plt.show()
```

As shown in the scree plot above, PCs 1 and 2 have eigenvalues greater than 1, while PCs 3, 4, and 5 fall at or below 1. Using the aforementioned Kaiser Rule as the guide, PCs 1 and 2 can be kept; the others can be discarded.

###  E3. Benefits

Principal Component Analysis reduces the dimensionality of a dataset by re-expressing larger numbers of potentially correlated variables as smaller numbers of linearly uncorrelated variables; this simplifies the data, which comes with a number of potential benefits. PCA generalizes the data in a way that can assist with machine learning models, helping to prevent overfitting the model to your initial dataset, and reducing the processing overhead of the ML algorithms. It also helps improve attempts at visualizing the data, as there are fewer dimensions to take into account when building the visualization. (Bigabid, n.d.; Soriano, n.d.).

Using the PCA performed above as an example, we could potentially replace Income, Tenure, MonthlyCharge, Bandwidth_GB_Year, and Outage_sec_perweek with PC1 and PC2; this would reduce the dimensionality of the data by replacing five of the variables with just two. The new PCs would act as generalizations of the replaced variables, allowing us to perform analysis on and draw conclusions from the data while considering fewer variables.

Machine learning models that could be built off of this new dataset using the PCs would be trained on these generalizations rather than the specific datapoints of the original dataset; this would help prevent overfitting of the model to the initial data and would allow greater accuracy when applying the model to new datasets. These models would also be more computationally efficient, as they would be running against a dataset of reduced size, and would therefore have to take three fewer variables into account. 

Similarly, we could build data visualizations to represent the data in a simplier way; we could, for example, consider displaying changes in other variables against these two PCs, rather than attempting to visualize them against all five of the different variables that we derived the PCs from.
