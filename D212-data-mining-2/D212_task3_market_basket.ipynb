### Part I: Research Question

#### A1. Research Question

The research question that will be explored in this paper is: can market basket analysis (MBA) be used to identify which items are frequently purchased together, when analyzing digital storefront transactions from a theoretical telecommunications company?

#### A2. Defined Goal

The goal of this analysis will be to utilize market basket analysis to identify items from the telecommunication company's storefront that are commonly purchased together. This will be achieved by analyzing storefront transactions that are stored in the teleco_market_basket.csv file. 

### Part II: Technique Justification

#### B1. Explanation of Market Basket

Market basket analysis (MBA) is an unsupervised machine learning method commonly used in retail analytics that helps analyze purchasing patterns. Specifically, MBA analyzes combinations of products that are often purchased together by identifying patterns within frequently purchased items by customers. (GeeksForGeeks, 2025).

For this specific analysis, the MBA method employed will input a dataset containing customer transactions and the items that were purchased in those transactions, then compare the items contained within those transactions against other items both within the same transaction and across other transactions. The expected outcome of this analysis is the creation of a listing  containing all item pairings along with metrics that indicate how likely it is that the items within these pairings are purchased together in the same transaction.

Market basket analysis assumes that items that are frequently bought together are related in some way, and that the relationship between these items can be projected onto future customer behavior in order to predict purchasing patterns. (Sivek, 2020)

### Part III: Data Preparation and Analysis

#### C1. Transforming the Dataset

Before performing a market basket analysis, it's necessary to clean and prepare the data. The first step will be to remove any rows or columns that are completely null or marked as NaN, as we are not interested in either empty transactions, or items that are not featured in any transaction. The code below imports the dataset into a pandas dataframe, then removes these empty rows and columns. The result is a cleaned dataframe called teleco_mb_cleaned_df, which contains 7501 rows and 20 columns.

```python
# Import additional packages for the analysis

import numpy as np
from mlxtend.preprocessing import TransactionEncoder
```

```python
# Get info about the teleco market basket dataset that we imported earlier
teleco_mb_df.info()
```

```python
# Clean the teleco market basket dataframe

# Place in new dataframe to maintain original input file integrity
teleco_mb_clean_df = teleco_mb_df.copy(deep=True)

# Check for and remove blank rows
teleco_mb_clean_df.dropna(axis=0,how='all',inplace=True)

# Check for and remove blank columns
teleco_mb_clean_df.dropna(axis=1,how='all',inplace=True)

# Get info about the newly-cleaned teleco market dataset
teleco_mb_clean_df.info()
```

The input dataset has now been successfully cleaned of empty columns and rows, and placed into the teleco_mb_cleaned_df dataframe. Data transformation is now required to prepare this dataframe for performing the market basket analysis. This involves creating a so-called "list of lists," or list of values contained within the dataset, and then using TransactionEncoder from the mlxtender package to transform the dataframe into a NumPy array that uses this list for column values. After generating this NumPy array, the data will then be placed back into a new pandas dataframe called teleco_mb_transaction_df. Any columns with all NaN values will then be dropped, resulting in the final dataframe that will be used for the MBA, clean_teleco_mb_transaction_df.

```python
# Perform data transformation to prepare for analysis

# Create list of lists
listolists = []
for i in range (0,7501):
    listolists.append([str(teleco_mb_clean_df.values[i,n])
for n in range (0,20)])
```

```python
# Feed list to TransactionEncoder; this creats a numpy boolean array

encoder = TransactionEncoder()
teleco_mb_array = encoder.fit(listolists).transform(listolists)

teleco_mb_array
```

```python
teleco_mb_transaction_df = pd.DataFrame(teleco_mb_array, columns = encoder.columns_)
```

```python
teleco_mb_transaction_df
```

```python
# Drop columns where all values are false
clean_teleco_mb_transaction_df = teleco_mb_transaction_df.drop(['nan'],axis=1)
clean_teleco_mb_transaction_df
```

```python
# Output cleaned/prepared dataset
```

```python
clean_teleco_mb_transaction_df.to_csv('clean_teleco_mb_transaction_df.csv', index=False)
```

#### C2. Code Execution

The code below utilizes the apriori and association_rules functions from the mltxend Python library to generate an association rules table, which describes the relationships between itemsets in the dataset.

```python
# Import additional packages for use in the analysis
from mlxtend.frequent_patterns import apriori, association_rules

# Create Apriori object
# Anything with support less than 0.02 will not be incuded
freq_items = apriori(clean_teleco_mb_transaction_df, min_support = 0.02, use_colnames = True)
freq_items
```

```python
# Association rules table
# Minimum threshold for lift set to 1, so any rule less than 1 will be excluded
# Threshold value of lift >= 1 is generally considered a strong relationship,
# indicating that these items are likely to be purchased together
rule_table = association_rules(freq_items, metric = 'lift', min_threshold = 1)
rule_table
```

#### C3. Association Rules Table

The association rules table has been printed below, including values for lift, support, and confidence.

```python
# Print rules table generated above
rule_table
```

#### C4. Top Three Rules

The first rule listed here has the highest value for the confidence metric; itcontains the antecedent "10ft iPHone Charger Cable 2 Pack" and the consequent "Dust-Off Compressed Gas 2 pack". The confidence value for this rule is 0.456464, which indicates that, out of all transactions containing iPhone Charger Cable 2 Pack, 45.65% of them also had Dust-Off Compress Gas 2 pack.

```python
# Rule with the highest confidence value
top_three_conf = rule_table.sort_values('confidence',ascending=False).head(1)
top_three_conf
```

The second rule, listed below, has the highest lift value out of all records on the association rules table. This rule contains VIVO Dual LCD Monitor Desk mount as the antecedent, and SanDisk Ultra 64GB card as the consequent. The lift value for this rule is 2.291162, which shows that customers are much more likely to purchase these items together, rather than separately.

```python
# Rule with the highest lift value
top_three_lift = rule_table.sort_values('lift',ascending=False).head(1)
top_three_lift
```

The third rule has the highest support value of any item pairing in the association rules table; it contains VIVO Dual LCD Monitor Desk mount as the antecedent, and the Dust-Off Compressed Gas 2 pack as teh consequent. The support value for this rule is 0.059725; this indicates that around 6% of all transactions contain both of these items.

```python
# Rule with the highest support value
top_three_support = rule_table.sort_values('support',ascending=False).head(1)
top_three_support
```

### Part IV: Data Summary and Implications

#### D1. Significance of Support, Lift, and Confidence Summary

There are three core metrics to evaluate when looking at association rules: support, confidence, and lift:
- Support reflects the proportion of all transactions that contain a given rule. This can give insight into how common a given item pairing is within the dataset. 
- Confidence is the proportion of all transactions in the dataset that contain all of the items in the transaction's itemset, divided by the proportion of transactions that contain just one of the items. This provides perspective on how common it is to see two items paired together, when compared with all transactions containing one of these items. 
- Lift measures the likelihood that two items will be purchased together, when compared with the likelihood that they will be purchased on their own. A lift value that is greater than 1 indicates that the items in a given pair are more likely to be purchased as part of that pair, versus being purchased individually and not together. (Sivek, 2020)

The rule with the highest confidence level is 10ft iPhone Charger Cable 2 Pack -> Dust-Off Compressed Gas 2 pack, with a confidence value of 0.456464. This means that, out of all transactions containing iPhone Charger Cable 2 Pack, 45.65% of them also had Dust-Off Compress Gas 2 pack.

The rule with the highest support value are VIVO Dual LCD Monitor Desk mount -> Dust-Off Compressed Gas 2 pack. The support value for these rules is 0.059725; this indicates that around 6% of all transactions contain both of these items.

The rule with the highest lift value contains the items VIVO Dual LCD Monitor Desk mount and SanDisk Ultra 64GB card. The lift value for this rule is 2.291162. This indicates that customers are much more likely to purchase these items together, implying a strong relationship between these items.
